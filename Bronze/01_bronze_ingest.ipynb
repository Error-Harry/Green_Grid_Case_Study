```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
dbutils.widgets.text("run_id", "manual")
run_id = dbutils.widgets.get("run_id")
```

```python
base_path = "/Volumes/databricksbootcamp/greeengridcasestudy/greeengridcasestudyvolume"
```

```python
df_pii = spark.read.csv(
        f"{base_path}/customer_pii.csv", 
        header=True
)

df_pii = (
        df_pii.withColumn("ingested_at", current_timestamp())
        .withColumn("run_id", lit(run_id))
)

(
df_pii.write
    .mode("overwrite")
    .saveAsTable("databricksbootcamp.greeengridcasestudy.bronze_pii")
)
```

```python
df_pii_test = spark.read.table("databricksbootcamp.greeengridcasestudy.bronze_pii")
```

```python
df_plans = spark.read.csv(
        f"{base_path}/pricing_plans.csv", 
        header=True,
        inferSchema=True
)

df_plans = (
        df_plans.withColumn("ingested_at", current_timestamp())
        .withColumn("run_id", lit(run_id))
)

(
df_plans.write
    .mode("overwrite")
    .saveAsTable("databricksbootcamp.greeengridcasestudy.bronze_plans")
)
```

```python
df_plans_test = spark.read.table("databricksbootcamp.greeengridcasestudy.bronze_plans")
```

```python
reading_schema = StructType([
    StructField("kwh", DoubleType(), True),
    StructField("meter_id", StringType(), True),
    StructField("status", StringType(), True),
    StructField("timestamp", TimestampType(), True)

])

df_readings = spark.read.json(
        f"{base_path}/meter_readings.json",
        multiLine=True,
        schema=reading_schema
)

df_readings = (
        df_readings.withColumn("ingested_at", current_timestamp())
        .withColumn("run_id", lit(run_id))
)

(
df_readings.write
    .mode("overwrite")
    .saveAsTable("databricksbootcamp.greeengridcasestudy.bronze_readings")
)
```

```python
df_readings_test = spark.read.table("databricksbootcamp.greeengridcasestudy.bronze_readings")
```
